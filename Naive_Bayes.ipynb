```python
# Import các thư viện cần thiết

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Tải bộ dữ liệu Iris


```python
from sklearn.datasets import load_iris

# Tải bộ dữ liệu Iris
iris = load_iris()

print("Tên các đặc trưng:")
print(iris.feature_names)

print("\nTên các lớp:")
print(iris.target_names)
```

## 2. Khám phá bộ dữ liệu

```python
# Chuyển dữ liệu Iris thành DataFrame

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Thêm cột nhãn (loài hoa)
df["species"] = iris.target

df.head()
```

```python
# Kiểm tra thông tin bộ dữ liệu

df.info()
```

## 3. Kiểm tra dữ liệu

```python
# Kiểm tra dữ liệu bị thiếu

df.isnull().sum()
```
